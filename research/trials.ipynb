```python
from langchain_community.document_loaders import PyPDFLoader, DirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Extract text from PDF files
def load_pdf_files(data):
    loader = DirectoryLoader(
        data,
        glob="*.pdf",
        loader_cls=PyPDFLoader
    )
    
    documents = loader.load()
    return documents
    
```

```python
extracted_data = load_pdf_files("data")
```

```python
from typing import List
from langchain_core.documents import Document

def filter_to_minimal_docs(docs: List[Document]) -> List[Document]:
    """
        Given a list of Document objects, return a new list of Document objects
        containing only 'source' in metadata and the original page_content.
    """
    
    minimal_docs: List[Document] = []
    for doc in docs:
        src = doc.metadata.get("source")
        minimal_docs.append(
            Document(
                page_content=doc.page_content,
                metadata={"source": src}
            )
        )
    
    return minimal_docs
```

```python
minimal_docs = filter_to_minimal_docs(extracted_data)
```

```python
minimal_docs
```

```python
# Split the documents into smaller chunks
def text_splitter(minimal_docs):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=500, # each chunk has 500 tokens
        chunk_overlap=20,
        length_function=len
    )
    texts_chunk = text_splitter.split_documents(minimal_docs)
    return texts_chunk
```

```python
texts_chunk = text_splitter(minimal_docs)
print(f"Number of chunks: {len(texts_chunk)} ")
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

def download_embeddings():
    """
    Download and return the HuggingFace embeddings model.
    """
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
    embeddings = HuggingFaceEmbeddings(
        model_name=model_name
    )
    
    return embeddings
embeddings = download_embeddings()
```

```python
# Creates a retriever object from the vector store (docsearch) so the application can retrieve the most relevant documents for a query
retriever = docsearch.as_retriever(search_type="similarity", search_kwargs={"k":3}) 
# as_retriever() converts the vector store into a retriever interface used by LangChain for RAG (Retrieval-Augmented Generation). Instead of manually searching vectors, the retriever handles embedding the query; searching the vector database; returning the most relevant documents
# search_type tells the retriever how to search. "similarity" returns the closest matches. Other keywords: "mmr" - Maximal Marginal Relevance; "similarity_score_threshold"
# search_kwargs controls how many documents are returned
# How it works in a Chatbot: for example, the user asks:"What does a family doctor do?". The retriever will convert the question into embedding, then search Pinecone for similar vectors and returns 3 relevant chunks of text. After that, those chunks are sent to the LLM to generate the answer.
```

```python
# Another way to retrieve the data by using a retriever
retrieved_docs = retriever.invoke("What is Acne?")
retrieved_docs
```

```python
# Connect LLM
from langchain_google_genai import ChatGoogleGenerativeAI

chatModel = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
from langchain_classic.chains import RetrievalQA
from langchain_core.prompts import ChatPromptTemplate
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_classic.chains import create_retrieval_chain
```

```python
# Create a system instruction for the AI model
# System prompt tells the AI: what role it should play; how it should respond; what rules it must follow
system_prompt = (
    "You are an medical assistant for question-answering tasks."
    "Use the following pieces of retrieved context to answer "
    "the question. If you don't know the answer, say that you "
    "don't know. Use three sentences maximum and keep the "
    "answer concise."
    "\n\n"
    "{context}"
)

# Create a structured chat prompt used by the LLM
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt), # tell AI how to behave
        ("human", "{input}"), # user's questions
    ]
)
```

```python
question_answer_chain = create_stuff_documents_chain(chatModel, prompt) # create a chain that sends retrieved documents to the LLM together with the prompt
rag_chain = create_retrieval_chain(retriever, question_answer_chain) # Combines retriever and LLM chain into a full RAG pipeline
```

```python
# Test the pipeline
response = rag_chain.invoke({"input": "what is acebutolol?"})
print(response["answer"])
```
